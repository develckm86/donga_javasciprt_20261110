```python
arr=[11,22,33,44]
for n in arr : #js for (let n of arr)
    print(n)
```

```python
user={"name":"경민","age":40}
for key in user : #js for( let key in user )
    print(key,user[key])
```

```python
print(user.items()) #js Object.entries(user)
for k,v in user.items() : #js for(let[k,v] of Object.entries(user))
    print(k,v)
```

```python
#언팩킹 (==js 구해분해할당)
nums=[10,11,12,13]
a,b,c,d=nums #let[a,b,c,d]=nums
print(a,b,c,d)
a,b,*rest=nums #let[a,b,...rest]=nums
print(a,b,rest)

#객체의 구조분해할당은 파이썬에서 불가능
name,age={"name":"경민","age":40}
print(name,age)
```
