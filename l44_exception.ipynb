```python
# with 컨텍스트관리자 (생성, 열기,해제)
import json
try :
    with open("movie.json","r",encoding="utf-8") as f :
        data=f.read()
        print(data)
        obj=json.loads(data)
except FileNotFoundError as e :
    print("파일 이름을 확인하세요!~",e)
except json.JSONDecodeError as e :
    print("json 형식이 아닙니다~!",e)
```

```python
from pathlib import Path #open처럼 파일을 불러옴 (with x)
import json
#예외처리하세요~
fileName="movie.json"
try :
    file = Path(fileName)
    data = file.read_text(encoding="utf-8")
    print(data)
    obj=json.loads(data)
except FileNotFoundError as e :
    print(f"{fileName}을 찾을 수 없습니다!")
except json.JSONDecodeError as e :
    print("json 형식이 잘못되었습니다.")
print("프로그램이 종료되지 않고 예외를 처리함")
```

```python
age="13"
try :
    age=int(age)
    print("성인" if age>20 else "미성년")
except ValueError as e :
    print("나이는 수만 입력하세요!")
```

```python
#성적을 입력하면 재시험 or 통과를 출력하는 프로그램
while True :
    score = input("성적을 입력하세요 :")
    #오류가 있으면 "성적은 수만 입력" 메세지를 보내고 
    #오류가 없다면 50점 이상일때 통과 미만일때 재시험을 안내하세요!
    try :
        score = int(score)
        msg="통과" if score>=50 else "재시험"
        print(msg)
    except ValueError as e :
        print("성적은 수만 입력")
```
