```python
import json #문자열 json을 dict으로 변환하는 라이브러리(모듈,도구)

student={"name":"둘리","age":30000,"isStudent":False,"father":None} #dict

studentJson='{"name":"길동","age":30,"isStudent":true,"father":null}' 
#str json(js기준): True=>true None=>null

print(student["name"])
#print(studentJson["name"]) #TypeError
gildong=json.loads(studentJson) #js JSON.parse(studentJson)
print(gildong["name"],gildong["age"],gildong["isStudent"],gildong["father"])
#ensure_ascii 문자열 바이트로 표현
stuStr=json.dumps(student,ensure_ascii=False,indent=4)
print(stuStr)
```
