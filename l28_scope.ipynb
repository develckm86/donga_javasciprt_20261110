```python
a=10
#    b=20 #들여쓰기가 지역scope, if while def 없이 들여쓰기 불가능
if a>=10 : #지역
    if a==10 :
        c=300
    b=20
    print(a,b,c) #파이썬의 변수는 if while 로 scope를 구분하지 않는다.
    if b>=20 : #자식지역
        c=30
        print(a,b,c)
print(b,c) #파이썬은 if while 의 들여쓰기를 완전한 scope로 인정하지 않는다
```

```python
x=10
def method(z=30) :
    y=20
    print(x,y,z)
#print(y) #함수 scope에 존재하는 변수는 함수가 실행될때만 존재한다.
method()
method(300)
method(400)
#파이썬은 변수의 scope를 함수에서만 사용가능 (들여쓰기가 scope지만 적용 안됨!)
```
