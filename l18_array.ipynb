```python
arr=["올드보이","기생충","왕사남"]
print(arr[0],arr[1],arr[2])
#print(arr[3]) #IndexError   
print(arr[-1])  #음수인덱싱
print(arr[1:3]) #슬라이싱(파이썬의고유기능!!!)
```

```python
import time
while True : #python 은 들여쓰기를 한 부분이 같은 영역(scope)
    print("안녕")
    time.sleep(1)
```

```python
#for : 반복할 횟수를 정하고 반복실행
#for(let i=0; i<5; i++){}
for i in range(0,5) : #[0,1,2,3,4]
    print(f"안녕 {i}번째")
```
