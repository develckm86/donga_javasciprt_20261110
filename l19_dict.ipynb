```python
#dict dictionary : 자료를 저장하는 data
movie={
    "name":"기생충",
    "director":"봉준호"
    }
print(movie["name"])
print(movie["director"])
movie["개봉일"]=2019
movie.update({"a":10,"b":True}) #js Object.assign(movie,{a:10,b:True})
movie.pop("b")
print(movie)
```

```python
print(movie.keys())
print(movie.values())
print("개봉일" in movie)
print(movie.items()) # [[key,value] ...] entry
print(type(movie)) #<class 'dict'>
```
