```python
def sum(a, b, c) :
    result=a+b+c
    return result
s=sum(10,20,30)
s2=sum(11,22,33)
print(s)
print(s2)
```

```python
def say(name) :
    print(f"저는 {name} 입니다~ 안녕하세요~!!")
say("경만")
result=say("경민")
print(result)
```

```python
#가변인자 : 매개변수를 복수로 받는 방법 (tuple,배열과유사)
def sum2(*args) :
    print(args)
    result=0
    for n in args :
        result+=n
    return result
s=sum2(10,11,12,13,14,15,16,17,18)
print(s)
```
